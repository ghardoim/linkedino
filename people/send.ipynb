```python
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By

from os.path import dirname, abspath
from sys import path

SCRIPT_DIR = dirname(abspath(__name__))
path.append(dirname(SCRIPT_DIR))

from utils.browser import Linkedin0, URLs
```

```python
browser = Linkedin0(URLs.SEARCH)
```

```python
search = browser.findby_classname("search-global-typeahead__input")

browser.clear(search)
search.send_keys("talent acquisition" + Keys.ENTER)
[browser.clickby_text(btn) for btn in ["People", "2nd", "3rd+"]]

browser.clickby_text("Locations")
location = browser.findby_xpath("//input[@placeholder='Add a location']")

for country in ["Estonia"]:
    browser.clear(location)
    location.send_keys(country)

    browser.wait_by(By.XPATH, "//div[@role='listbox']")
    location.send_keys(Keys.ARROW_DOWN)
    location.send_keys(Keys.ENTER)

browser.clickby_text("Show results", "span")

for person in browser.findby_xpath("//ul[@role='list']").find_elements(By.TAG_NAME, "li"):
    if "Connect" not in person.text: continue

    browser.clickby_text("Connect", "span")
    browser.wait_by(By.XPATH, "//div[@role='dialog']")

    browser.clickby_text("Send without a note", "span")
    browser.wait_not(By.XPATH, "//div[@role='dialog']")
```

```python
browser.logout()
```
